```python
# Filename: group_pipeline.ipynb

# --- 1. IMPORTS & SETUP ---
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler

# Mount your Google Drive to access files
from google.colab import drive
drive.mount('/content/drive')

# Load data (use the same code from Step 1)
file_path = '/content/drive/MyDrive/AI_ML/Dataset/winequality-red.csv'
df = pd.read_csv(file_path)
print("Initial shape:", df.shape)


# --- 2. HANDLING MISSING DATA (Task 1) ---
# (Add code for checking and handling missing data if there was any)
# Since there's none, we can add a comment explaining it was checked.
print("\nStep 1: Missing value check complete. No missing values found.")


# --- 3. FEATURE ENGINEERING (Task 5) ---
# It's often good to create features before removing outliers or scaling
df['total acidity'] = df['fixed acidity'] + df['volatile acidity']
print("\nStep 2: 'total acidity' feature created.")


# --- 4. OUTLIER REMOVAL (Task 2) ---
features_to_check = df.drop('quality', axis=1).columns
for col in features_to_check:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    df = df[(df[col] >= lower_bound) & (df[col] <= upper_bound)]
print(f"\nStep 3: Outliers removed. New shape: {df.shape}")


# --- 5. SEPARATE FEATURES AND TARGET ---
X = df.drop('quality', axis=1)
y = df['quality']


# --- 6. DATA SCALING (Task 3) ---
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled_df = pd.DataFrame(X_scaled, columns=X.columns)
print("\nStep 4: Features have been scaled.")


# --- 7. FINAL PROCESSED DATA ---
print("\nPreprocessing complete. Here are the first 5 rows of the final processed features:")
X_scaled_df.head()

# Now X_scaled_df (features) and y (target) are ready for model training!
```
